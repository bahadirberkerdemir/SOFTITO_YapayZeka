```python
notlar =[]
for i in range(1, 5):
    notun = float(input(f"{i}. dersin notunu giriniz: "))
    notlar.append(notun)

ortalama = sum(notlar) / len(notlar)
maxx = max(notlar)
minn = min(notlar)
if ortalama >= 90:
    harf_notu = "AA"
elif ortalama >= 85:
    harf_notu = "BA"
elif ortalama >= 80:
    harf_notu = "BB"
elif ortalama >= 75:
    harf_notu = "CB"
elif ortalama >= 70:
    harf_notu = "CC"
elif ortalama >= 65:
    harf_notu = "DC"
elif ortalama >= 60:
    harf_notu = "DD"
elif ortalama >= 50:
    harf_notu = "FD"
else:
    harf_notu = "FF"

print(f"Not Ortalaması : {ortalama}")
print(f"Harf Notu      : {harf_notu}")
print(f"En Yüksek Not  : {maxx}")
print(f"En Düşük Not   : {minn}")
```

```python
sepet = []

for i in range(1, 6):
    fiyat = float(input(f"{i}. ürünün fiyatını giriniz: "))
    sepet.append(fiyat)

pahali_urunler = [fiyat for fiyat in sepet if fiyat > 100]

toplam_tutar = sum(sepet)

print("\n100 TL üzeri ürünler:", pahali_urunler)
print(f"Sepet Toplamı: {toplam_tutar} TL")

if toplam_tutar > 500:
    indirimli_tutar = toplam_tutar * 0.90
    print("%10 indirim kazandınız!")
    print(f"İndirimli Ödenecek Tutar: {indirimli_tutar} TL")
else:
    print(f"Ödenecek Tutar: {toplam_tutar} TL")
```
